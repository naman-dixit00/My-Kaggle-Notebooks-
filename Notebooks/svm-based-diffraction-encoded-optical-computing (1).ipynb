**Overview**
Photonic computation and optical signal classification rely fundamentally on the physics of wave propagation. # Scalar Diffraction Theory
**Theory**
Under the scalar approximation (valid when polarization effects are negligible), an optical field can be represented as a complex scalar function:  

> [U(x, y, z) = A(x, y, z) , e^{i\phi(x, y, z)}]
  
where:  
•	(A(x, y, z)) is the amplitude  
•	(\phi(x, y, z)) is the phase  
Free-space propagation of this field from plane (z = 0) to (z = d) is governed by the Huygens–  Fresnel principle, which in Fourier form yields the Fresnel diffraction integral :  

> [U_d(x, y) = \mathcal{F}^{-1} \left[ \mathcal{F}{U_0(x, y)}\cdot H(f_x, f_y)\right]]  

where the transfer function is:  

> [  H(f_x, f_y) =  \exp\left(-i \pi \lambda d (f_x^2 + f_y^2)\right)  ]  

This formulation is computationally efficient and preserves phase coherence—essential for interference modeling.  


# Computational Representation of Optical Fields
**Theory**
In numerical simulations, the continuous optical field is discretized on a 2D grid:
•	Each pixel represents a spatial sample of the complex field
•	The field is stored as a complex-valued matrix (U \in \mathbb{C}^{N \times N})
•	Observed optical signals correspond to intensity patterns:

> [I(x, y) = |U(x, y)|^2]

These intensity matrices form the diffraction-encoded optical data used for classification.
________________________________________


# Fourier-Domain Propagation Operator
**Theory**  
The Fresnel propagation operator can be implemented numerically using FFTs:
1.	Compute spatial frequency coordinates
2.	Construct the propagation transfer function
3.	Apply Fourier transform → multiply → inverse transform
This operator is linear and unitary (up to sampling), preserving energy and enabling physically meaningful transformations.


```python
# ================================
# Fresnel Diffraction Propagation
# ================================

def fresnel_propagation(field, wavelength, pixel_size, distance):
    """
    Fresnel propagation of a complex optical field using Fourier optics.

    Parameters:
    field : np.ndarray (complex)
        Input optical field at z = 0
    wavelength : float
        Optical wavelength (meters)
    pixel_size : float
        Spatial sampling interval (meters)
    distance : float
        Propagation distance (meters)

    Returns:
    propagated_field : np.ndarray (complex)
        Optical field after propagation
    """
    N = field.shape[0]

    # Spatial frequency coordinates
    fx = np.fft.fftfreq(N, d=pixel_size)
    fy = np.fft.fftfreq(N, d=pixel_size)
    FX, FY = np.meshgrid(fx, fy)

    # Fresnel transfer function
    H = np.exp(-1j * np.pi * wavelength * distance * (FX**2 + FY**2))

    # Fourier-domain propagation
    field_ft = fft.fft2(field)
    propagated_field = fft.ifft2(field_ft * H)

    return propagated_field
```

#  Generation of Interference Patterns
**Theory**  
To simulate high-speed optical interference, we construct input fields with:
•	Uniform amplitude
•	Spatially varying random phase masks
Multiple phase realizations produce distinct interference geometries after propagation. These geometries define separable or non-separable regions in feature space, which the SVM will later learn to classify.

---
Code: Phase-Encoded Optical Field Generator
---

```python
# ================================
# Optical Field Generator
# ================================

def generate_phase_encoded_field(grid_size):
    """
    Generate a phase-encoded coherent optical field.

    Parameters:
    grid_size : int
        Spatial resolution of the optical field

    Returns:
    field : np.ndarray (complex)
        Phase-encoded optical field
    """
    amplitude = np.ones((grid_size, grid_size))
    phase = 2 * np.pi * np.random.rand(grid_size, grid_size)

    field = amplitude * np.exp(1j * phase)
    return field
```

# Simulation of Diffraction-Encoded Intensity Images
**Theory**  
The physically measurable quantity is the intensity, not the complex field. After propagation, intensity is computed as:

> [I(x, y) = |U_d(x, y)|^2]

These intensity matrices constitute the High-Speed Optical Interference Simulation Images used as the dataset for classification.

---
Code: Single Interference Pattern Simulation
---

```python
# ================================
# Interference Pattern Simulation
# ================================

def simulate_interference_pattern(config):
    """
    Simulate a single diffraction-based optical interference pattern.

    Parameters:
    config : dict
        Optical simulation configuration

    Returns:
    intensity : np.ndarray
        Normalized intensity image
    """
    field = generate_phase_encoded_field(config["grid_size"])
    propagated = fresnel_propagation(
        field,
        wavelength=config["wavelength"],
        pixel_size=config["pixel_size"],
        distance=config["propagation_distance"]
    )

    intensity = np.abs(propagated)**2
    intensity /= intensity.max()  # normalization

    # Add weak sensor noise
    intensity += config["noise_level"] * np.random.randn(*intensity.shape)

    return intensity
```

# High-Dimensional Feature Embedding of Light-Field Manifolds
**Overview**  
Diffraction-encoded optical interference patterns are not arbitrary images; they reside on structured, physics-constrained manifolds induced by wave propagation, coherence, and phase superposition. Direct classification in pixel space obscures this geometry and leads to suboptimal margin formation.
This section constructs high-dimensional feature embeddings that preserve physically meaningful correlations in light-field space while enabling efficient convex margin optimization via Support Vector Machines (SVMs). The objective is to transform raw optical intensity matrices into feature vectors that retain diffraction geometry, interference structure, and propagation-induced dependencies.

---

# Light-Field Space as a High-Dimensional Manifold
**Theory**  
Each diffraction-encoded intensity image  
[
I(x, y) = |U_d(x, y)|^2
]  
can be interpreted as a point in a high-dimensional space:  
[
\mathcal{X} \subset \mathbb{R}^{N^2}  
]  
However, due to physical constraints (wave equation, coherence, finite   aperture), valid optical fields occupy a low-dimensional nonlinear manifold     embedded within this ambient space.    
Key properties of this manifold:    
•	Strong spatial correlations  
•	Energy conservation constraints  
•	Phase-induced interference structure  
•	Smoothness under propagation distance variations  
The embedding strategy must preserve these properties.  


# Vectorization with Physical Normalization
**Theory**  
The first embedding step converts each 2D optical field into a vector while enforcing physically meaningful normalization:  
•	Intensity normalization → energy invariance  
•	Mean-centering → removes global illumination bias  
•	Scaling → ensures margin geometry is not dominated by amplitude  
Given an intensity image (I \in \mathbb{R}^{N \times N}) :  
[
\mathbf{x} = \text{vec}(I) \in \mathbb{R}^{N^2}
]  
This vector represents a point in light-field feature space.  

---
Code: Dataset Generation and Vectorization
---

```python

# ================================
# Generate Optical Dataset
# ================================

def generate_optical_dataset(config, labels_rule="phase_parity"):
    """
    Generate a labeled dataset of diffraction-encoded optical patterns.

    Parameters:
    config : dict
        Optical simulation parameters
    labels_rule : str
        Rule used to assign class labels (physically motivated)

    Returns:
    X : np.ndarray
        Flattened optical feature vectors
    y : np.ndarray
        Class labels
    """
    X = []
    y = []

    for _ in range(config["num_samples"]):
        field = generate_phase_encoded_field(config["grid_size"])
        propagated = fresnel_propagation(
            field,
            config["wavelength"],
            config["pixel_size"],
            config["propagation_distance"]
        )

        intensity = np.abs(propagated)**2
        intensity /= intensity.max()

        # Feature vector
        feature_vec = intensity.flatten()
        X.append(feature_vec)

        # Physically motivated labeling rule
        # Example: global phase-energy distribution asymmetry
        label = int(np.mean(intensity[:config["grid_size"]//2]) >
                    np.mean(intensity[config["grid_size"]//2:]))
        y.append(label)

    return np.array(X), np.array(y)

```

# Maximum Margin Principle in Light-Field Space
**Theory**  
* **Given a labeled dataset**  
[
{(\mathbf{x}i, y_i)}{i=1}^N,\quad \mathbf{x}_i \in \mathbb{R}^{D},; y_i \in {-1, +1}
]  
* **the SVM seeks a hyperplane:**
[
\mathbf{w}^\top \mathbf{x} + b = 0
]  
* **that maximizes the geometric margin:**
[
\gamma = \frac{2}{|\mathbf{w}|}
]  
* **subject to:**  
[
y_i(\mathbf{w}^\top \mathbf{x}_i + b) \geq 1 - \xi_i
]  
* **In photonic terms:**  
•	(\mathbf{x}_i): diffraction-encoded optical field configuration  
•	(\mathbf{w}): learned photonic decision normal  
•	Margin: tolerance of optical computation to field perturbations  


# Soft-Margin SVM for Optical Interference Data
**Theory**  
Optical interference patterns are inherently noisy due to:  
•	Finite coherence length  
•	Sensor noise  
•	Phase randomness  
Therefore, a soft-margin SVM is required. The optimization becomes:  
[
\min_{\mathbf{w}, b, \xi} ;\frac{1}{2}|\mathbf{w}|^2 + C \sum_i \xi_i
]  
where (C) controls the trade-off between:  
•	Margin width (photonic robustness)  
•	Classification fidelity (optical separability)

# Kernel Interpretation as Optical Transfer Functions
---
**Overview**  
Linear SVMs assume that diffraction-encoded optical manifolds are linearly separable in the embedded light-field space. However, realistic optical systems exhibit nonlinear separability arising from multi-path interference, higher-order diffraction effects, and phase–amplitude coupling. To model such phenomena, we introduce kernelized Support Vector Machines, where kernels act as implicit optical transfer functions that lift light-field representations into higher-dimensional reproducing kernel Hilbert spaces (RKHS).  
In this section, we establish a physics-informed interpretation of SVM kernels as analogs of optical propagation, scattering, and interference operators, thereby connecting statistical learning theory with photonic computation principles.


# Kernel Trick as Implicit Optical Propagation
**Theory**  
**A kernelized SVM replaces the inner product:**  
[
\langle \mathbf{x}_i, \mathbf{x}_j \rangle
]  
**with a kernel function:**  
[
K(\mathbf{x}_i, \mathbf{x}_j) = \langle \Phi(\mathbf{x}_i), \Phi(\mathbf{x}_j) \rangle
]  
where (\Phi(\cdot)) is an implicit nonlinear feature map.  
**In photonic terms**:  
•	(\mathbf{x}): diffraction-encoded optical field  
•	(\Phi(\mathbf{x})): virtual optical propagation through a nonlinear medium  
•	Kernel: effective optical transfer function  
Thus, the kernel trick corresponds to computing interference in an unobserved photonic domain.  
________________________________________


# Radial Basis Function (RBF) Kernel as Diffraction Blur
**Theory**  
**The RBF kernel is defined as:**  
[
K_{\text{RBF}}(\mathbf{x}_i, \mathbf{x}_j)
\exp\left(-\gamma |\mathbf{x}_i - \mathbf{x}_j|^2\right)
]  
This kernel has a direct physical analogy:  
•	Gaussian decay ↔ spatial coherence envelope  
•	(\gamma) ↔ inverse diffraction spread  
•	Distance (|\mathbf{x}_i - \mathbf{x}_j|) ↔ field mismatch energy  
Interpretation:  
The RBF kernel behaves like a coherent diffraction blur, where optical fields   interfere strongly only within a finite coherence radius.  

---
Code: RBF Kernel SVM for Optical Interference Data
---

```python
# ================================
# RBF Kernel SVM (Photonic Nonlinearity)
# ================================

rbf_svm = SVC(
    kernel="rbf",
    C=5.0,
    gamma=1e-3
)

rbf_svm.fit(X_scaled, y)

print("RBF SVM trained successfully.")
print("Number of support vectors per class:", rbf_svm.n_support_)
```

# Polynomial Kernel as Multi-Path Interference Operator

---

**Theory**  
**The polynomial kernel is defined as:**  
[
K_{\text{poly}}(\mathbf{x}_i, \mathbf{x}_j)
(\mathbf{x}_i^\top \mathbf{x}_j + c)^d
]  
**This kernel implicitly models:**  
•	Higher-order feature interactions  
•	Multiplicative coupling between spatial regions  
Photonic interpretation:  
•	Degree (d) ↔ number of interfering optical paths  
•	Polynomial expansion ↔ cascaded diffraction and reflection events  
This kernel approximates multi-bounce photonic interference.  

---
Code: Polynomial Kernel SVM
---

```python
# ================================
# Polynomial Kernel SVM
# ================================

poly_svm = SVC(
    kernel="poly",
    degree=3,
    C=1.0,
    coef0=1.0
)

poly_svm.fit(X_scaled, y)

print("Polynomial SVM trained successfully.")
print("Number of support vectors per class:", poly_svm.n_support_)
```

---
**Theory**  
 * To compare how different kernels reshape photonic decision boundaries, we project their decision functions onto the PCA-reduced light-field manifold.          
 * Differences in curvature and smoothness reflect distinct photonic computation regimes.

Code: Decision Function Comparison
---

```python
import numpy as np
import scipy.fft as fft
import matplotlib.pyplot as plt
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# DATASET_CONFIG from YIHcFRf9w07s
DATASET_CONFIG = {
    "wavelength": 532e-9,
    "pixel_size": 6.5e-6,
    "grid_size": 128,
    "propagation_distance": 0.15,
    "num_samples": 2000,
    "noise_level": 0.01
}

# fresnel_propagation from sxZun_IPxSK8
def fresnel_propagation(field, wavelength, pixel_size, distance):
    N = field.shape[0]
    fx = np.fft.fftfreq(N, d=pixel_size)
    fy = np.fft.fftfreq(N, d=pixel_size)
    FX, FY = np.meshgrid(fx, fy)
    H = np.exp(-1j * np.pi * wavelength * distance * (FX**2 + FY**2))
    field_ft = fft.fft2(field)
    propagated_field = fft.ifft2(field_ft * H)
    return propagated_field

# generate_phase_encoded_field from UsMyMBPDxXE0
def generate_phase_encoded_field(grid_size):
    amplitude = np.ones((grid_size, grid_size))
    phase = 2 * np.pi * np.random.rand(grid_size, grid_size)
    field = amplitude * np.exp(1j * phase)
    return field

# generate_optical_dataset from 9iWeGia5JplL
def generate_optical_dataset(config, labels_rule="phase_parity"):
    X_data = []
    y_data = []
    for _ in range(config["num_samples"]):
        field = generate_phase_encoded_field(config["grid_size"])
        propagated = fresnel_propagation(
            field, config["wavelength"], config["pixel_size"], config["propagation_distance"]
        )
        intensity = np.abs(propagated)**2
        intensity /= intensity.max()
        feature_vec = intensity.flatten()
        X_data.append(feature_vec)
        label = int(np.mean(intensity[:config["grid_size"]//2]) > \
                    np.mean(intensity[config["grid_size"]//2:]))
        y_data.append(label)
    return np.array(X_data), np.array(y_data)

# Data generation and scaling from O1XHRNMOQrKe
X_raw, y = generate_optical_dataset(DATASET_CONFIG)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_raw)

# PCA for Manifold Visualization from ynk5w8AFQ9W_
pca = PCA(n_components=3)
X_pca = pca.fit_transform(X_scaled)

# Linear SVM from Nm9AtkJHYYYK
linear_svm = SVC(kernel="linear", C=1.0)
linear_svm.fit(X_scaled, y)

# RBF SVM from v7u0uv_ni-N8
rbf_svm = SVC(kernel="rbf", C=5.0, gamma=1e-3)
rbf_svm.fit(X_scaled, y)

# Polynomial SVM from KhxZxJWbl57T
poly_svm = SVC(kernel="poly", degree=3, C=1.0, coef0=1.0)
poly_svm.fit(X_scaled, y)

# ================================
# Kernel Decision Geometry Comparison
# ================================

decision_linear = linear_svm.decision_function(X_scaled)
decision_rbf = rbf_svm.decision_function(X_scaled)
decision_poly = poly_svm.decision_function(X_scaled)

plt.figure(figsize=(15, 4))

for i, (decision, title) in enumerate([
    (decision_linear, "Linear Kernel"),
    (decision_rbf, "RBF Kernel"),
    (decision_poly, "Polynomial Kernel")
]):
    plt.subplot(1, 3, i + 1)
    plt.scatter(
        X_pca[:, 0],
        X_pca[:, 1],
        c=decision,
        cmap="coolwarm",
        alpha=0.7
    )
    plt.title(title)
    plt.xlabel("PC 1")
    plt.ylabel("PC 2")
    plt.colorbar(label="Signed Distance")

plt.suptitle("Kernel-Induced Photonic Decision Geometry")
plt.tight_layout()
plt.show()
```

# Where the SVM Abstraction Breaks
---
Despite its strengths, the SVM framework imposes abstractions that do not fully capture physical optical systems.  
# Lack of True Optical Nonlinearity
---
SVM kernels simulate nonlinearity mathematically but do not represent actual **nonlinear light–matter interactions such as:**  
•	Kerr effects  
•	Saturable absorption  
•	Two-photon absorption  
Thus, kernel-induced nonlinear boundaries should be interpreted as effective photonic models, not direct physical simulations.  


# Absence of Temporal Dynamics

---

This work focuses on static interference snapshots. Real photonic computing systems may involve:
•	Time-varying phase modulation
•	Pulsed or ultrafast dynamics
•	Spatiotemporal coherence effects
SVMs lack an intrinsic temporal state, limiting their ability to model dynamic photonic computation.


# Finite-Dimensional Approximation of Infinite Optical Fields
---
Optical fields are continuous and infinite-dimensional. Discretization onto pixel grids introduces:  
•	Sampling artifacts  
•	Aliasing  
•	Boundary effects  
While sufficient for margin geometry analysis, this approximation limits fidelity when mapping directly to hardware.
